    - cold
        - components with 2-5 km/s dispersion
    - warm
        - components with 8-6 km/s dispersion
    - hot
        - components with 8-100 km/s dispersion
    - sgfit
        - single gaussian fit for all pixels
    - psgfit
        - single gaussian fit for pixels where the optimal n-gauss = 1
    - ngfit
        - all gaussian components output by baygaud.py, in original order
    - ngfit_wrt_integrated_int
        - gaussians sorted by integrated intensity
    - ngfit_wrt_peak_amp
        - gaussians components sorted by peak flux
    - ngfit_wrt_vdisp
        - gaussian components sorted by velocity dispersion
    - ngfit_wrt_vlos
        - gaussian components sorted by los velocity
